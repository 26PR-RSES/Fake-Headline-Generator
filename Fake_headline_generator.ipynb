```python
# import random module
import random
```

```python
#create a function for 1st category: General News
def general_headlines():
    #create list of subjects
    subjects = ["Shahrukh Khan",
                "Amitabh Bachhan",
                "Prime Minister",
                "President Donald Trump",
                "Mega Star Joseph Vijay",
                "Nirmala Tai"
    ]
    
    #create a list of actions
    actions = ["spotted dancing",
               "Abused",
               "launches",
               "Eating",
               "was palying",
               "mocked",
               "creates"
    ]
    
    #create a list of places_or_things
    places_or_thigns = ["Marina Beach",
                        "On a car",
                        "Himself",
                        "a cat on the roof",
                        "Dog lover",
                        "In the bathroom"
                       ]
    
    #Starting the while loop
    while True:
        #randomly choosing subejcts, actions and objects
        subject = random.choice(subjects)
        action = random.choice(actions)
        thing_or_place = random.choice(places_or_thigns)
        
        #concating the headlins using the f-string
        headline = f"Breaking News: {subject} {action} {thing_or_place}"
        print("\n", headline)
        
        #taking user input 
        user_input = input('Do you want to generate a new headline(yes/no):').strip().lower()
        if user_input == 'no':
            with open("Fake-Headline-Generator/headline.txt", 'a') as f:
                f.write(headline + '\n')
            break
    
```

```python
general_headlines()
```
